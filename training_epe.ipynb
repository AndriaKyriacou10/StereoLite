```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
training_csv = pd.read_csv('phase1_training_Jul14_07-55-48.csv')
training_csv
```

```python
epe = np.array(training_csv['Value'], dtype=np.float64)
step = np.array(training_csv['Step'], dtype=np.float64)
```

```python
index=131

plt.plot(step, epe, linewidth='2.0')
plt.grid(True)
plt.xlabel('Steps')
plt.ylabel('EPE')
plt.title('Training EPE')
plt.show()


plt.plot(step[index:], epe[index:], linewidth='2.0')
plt.grid(True)
plt.xlabel('Steps')
plt.ylabel('EPE')
plt.title('Training EPE')
plt.show()
```
